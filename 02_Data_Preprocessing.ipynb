# YO G News — 02 Data Preprocessing

This notebook prepares news text for machine-learning training.

**Steps**
1. Load the dataset
2. Remove unusable rows
3. Combine title and article content
4. Normalize basic text formatting
5. Prepare features (`X`) and target (`y`)
6. Split the data into training and testing sets


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

DATA_PATH = Path("../dataset/news_dataset.csv")
if not DATA_PATH.exists():
    DATA_PATH = Path("../YO_G_News_Starter_Dataset.csv")
if not DATA_PATH.exists():
    DATA_PATH = Path("YO_G_News_Starter_Dataset.csv")

print("Dataset path:", DATA_PATH)

from sklearn.model_selection import train_test_split
```

```python
df = pd.read_csv(DATA_PATH)

required = ["title", "content", "credibility_label"]
missing = [c for c in required if c not in df.columns]
if missing:
    raise ValueError(f"Missing required columns: {missing}")

df = df.dropna(subset=["credibility_label"]).copy()

df["title"] = df["title"].fillna("").astype(str)
df["content"] = df["content"].fillna("").astype(str)

df["text"] = (df["title"] + " " + df["content"]).str.replace(r"\s+", " ", regex=True).str.strip()

df = df[df["text"].str.len() > 0].drop_duplicates(subset=["text"]).reset_index(drop=True)

print("Rows after cleaning:", len(df))
display(df[["title", "content", "credibility_label", "text"]].head())
```

```python
X = df["text"]
y = df["credibility_label"]

print("Features:", X.shape)
print("Target:", y.shape)
print("\nClasses:")
print(y.value_counts())
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))
```

## Preprocessing Result

The cleaned `text` column combines the headline and article body. This gives the classifier both the headline signal and the article context.
