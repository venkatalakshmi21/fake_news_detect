# YO G News — 01 Data Exploration

This notebook explores the news dataset before model training.

**Goals**
- Load the dataset
- Understand columns and data types
- Check missing and duplicate records
- Inspect credibility-label distribution
- Examine article length


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

DATA_PATH = Path("../dataset/news_dataset.csv")
if not DATA_PATH.exists():
    DATA_PATH = Path("../YO_G_News_Starter_Dataset.csv")
if not DATA_PATH.exists():
    DATA_PATH = Path("YO_G_News_Starter_Dataset.csv")

print("Dataset path:", DATA_PATH)
```

```python
df = pd.read_csv(DATA_PATH)

print("Dataset shape:", df.shape)
display(df.head())
print("\nColumns:")
print(df.columns.tolist())
print("\nData types:")
print(df.dtypes)
```

```python
print("Missing values:")
display(df.isnull().sum().to_frame("missing_values"))

print("Duplicate rows:", df.duplicated().sum())
```

```python
label_col = "credibility_label"

if label_col in df.columns:
    print("Credibility label distribution:")
    display(df[label_col].value_counts().to_frame("count"))

    df[label_col].value_counts().plot(kind="bar")
    plt.title("Credibility Label Distribution")
    plt.xlabel("Credibility Label")
    plt.ylabel("Number of Articles")
    plt.xticks(rotation=30, ha="right")
    plt.tight_layout()
    plt.show()
else:
    print(f"Column '{label_col}' was not found.")
```

```python
if "content" in df.columns:
    df["content_length"] = df["content"].fillna("").astype(str).str.len()
    display(df["content_length"].describe())

    df["content_length"].plot(kind="hist", bins=20)
    plt.title("Article Content Length")
    plt.xlabel("Characters")
    plt.ylabel("Number of Articles")
    plt.tight_layout()
    plt.show()
```

```python
if {"title", "content", "category", "credibility_label"}.issubset(df.columns):
    display(df[["title", "category", "credibility_label"]].head(10))
```

## Observation

The exploration step helps identify dataset quality issues before preprocessing and training. The credibility labels are the target values used by the YO G News classifier.
